```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from matplotlib import pyplot as plt
from keras.datasets import fashion_mnist
from keras.models import Sequential
```

```python
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0
x_train = x_train.reshape([-1, 28*28])
x_test = x_test.reshape([-1, 28*28])
```

```python
model = Sequential()
model.add(keras.Input(shape=(28*28,)))
model.add(keras.layers.Dense(512, activation='relu'))
model.add(keras.layers.Dropout(0.3))
model.add(keras.layers.Dense(256, activation='relu'))
model.add(keras.layers.Dropout(0.3))
model.add(keras.layers.Dense(10, activation='softmax'))

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.0005),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.fit(x_train, y_train, batch_size=64, epochs=150, validation_data=(x_test, y_test))
```

## Висновок:
У ході виконання завдання була побудована повнозв'язна нейронна мережа для класифікації товарів з датасету fashion_mnist. Фінальна архітектура: вхідний шар (784 нейрони) → Dense(512, ReLU) → Dropout(0.2) → Dense(256, ReLU) → Dropout(0.2) → вихідний шар (10, Softmax). Оптимізатор — Adam з learning rate 0.0005, функція втрат — sparse_categorical_crossentropy, batch size — 64, 150 епох.
В процесі пошуку оптимальної архітектури я експериментував з різними гіперпараметрами. Збільшення кількості нейронів з 128 до 512 дало помітний приріст точності. Додавання Dropout(0.2) зменшило overfitting — без регуляризації розрив між training і validation accuracy був значно більшим. Зменшення learning rate з 0.001 до 0.0005 стабілізувало навчання і дало додатковий приріст val_accuracy. Також були протестовані BatchNormalization, L2 регуляризація, SGD з momentum та різні значення batch size, однак ці зміни не дали суттєвого покращення.
Найкращий результат на тестових даних склав 90.73% (епоха 145). В процесі навчання спостерігався overfitting — training accuracy досягала ~97% тоді як validation accuracy стабілізувалась на рівні ~90.5-90.7%. Подальші експерименти з регуляризацією дозволили зменшити коливання val_accuracy і підвищити стабільність навчання, однак суттєво збільшити val_accuracy не вдалось.
